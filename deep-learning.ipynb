```python

from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["flower_type"] = [
    iris.target_names[i] for i in iris.target
]

df.head()
```

```python

print("Rows and columns:", df.shape)
print("Column names:", df.columns.tolist())
print("Missing values:")
print(df.isnull().sum())
```
